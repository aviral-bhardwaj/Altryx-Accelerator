```python
import os
from pathlib import Path
IS_DATABRICKS = "dbutils" in globals()
def widget(name, default):
    if IS_DATABRICKS:
        dbutils.widgets.text(name, default, name)
        return dbutils.widgets.get(name)
    return os.environ.get("SKILL_" + name.upper(), default)
wheel_path = widget("wheel_path", "/Volumes/<catalog>/<schema>/altryx_accelerator/altryx_accelerator-1.1.0-py3-none-any.whl")
pip_fallback = widget("pip_fallback", "https://github.com/aviral-bhardwaj/Altryx-Accelerator/releases/download/v1.1.0/altryx_accelerator-1.1.0-py3-none-any.whl")
```

```python
# Notebook-scoped %pip installation; local tests use an already installed wheel.
if "dbutils" in globals():
    import base64, tempfile, importlib.metadata, importlib.util
    from pathlib import Path
    install_target = wheel_path if Path(wheel_path).is_file() else pip_fallback
    try:
        import shlex
        get_ipython().run_line_magic("pip", "install " + shlex.quote(install_target))
        if importlib.metadata.version("altryx-accelerator") != "1.1.0":
            raise RuntimeError("Release wheel installation did not complete")
    except Exception:
        # OFFLINE_WHEEL is replaced deterministically by marketplace/build_notebooks.py.
        OFFLINE_WHEEL = ""
        if not OFFLINE_WHEEL:
            raise RuntimeError("Upload the release wheel and its dependencies to a UC Volume, set wheel_path, then rerun.")
        local_wheel = Path(tempfile.gettempdir()) / "altryx_accelerator-1.1.0-py3-none-any.whl"
        local_wheel.write_bytes(base64.b64decode(OFFLINE_WHEEL))
        get_ipython().run_line_magic("pip", "install --no-deps " + str(local_wheel))
    missing = [name for name in ("lxml", "yaml", "nbformat") if importlib.util.find_spec(name) is None]
    if missing:
        raise RuntimeError("Install lxml, PyYAML and nbformat from an approved wheelhouse before running offline: " + str(missing))
    dbutils.library.restartPython()
```

```python
import json, logging, os
from pathlib import Path
IS_DATABRICKS = "dbutils" in globals()
def widget(name, default):
    if IS_DATABRICKS:
        dbutils.widgets.text(name, default, name)
        return dbutils.widgets.get(name)
    return os.environ.get("SKILL_" + name.upper(), default)
wheel_path = widget("wheel_path", "/Volumes/<catalog>/<schema>/altryx_accelerator/altryx_accelerator-1.1.0-py3-none-any.whl")
SAMPLES_DIR = Path(widget("samples_dir", str(Path(wheel_path).parent / "samples")))
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger("skill_mode")
from altryx_accelerator.parser import AlteryxWorkflowParser
from altryx_accelerator.converter_engine import ConverterEngine, load_tool_mapping
from altryx_accelerator.self_correction import SelfCorrectingConverter
from altryx_accelerator.databricks_exporter import DatabricksExporter, batch_export
from altryx_accelerator.validators import run_generated_code
```

```python
input_dir = widget("input_dir", str(SAMPLES_DIR))
output_dir = widget("output_dir", "/tmp/altryx_batch")
formats = tuple(widget("output_format", "ipynb,py").split(","))
source_file = widget("source_tables_config", "")
mappings = json.loads(Path(source_file).read_text()) if source_file else {}
engine = ConverterEngine(source_tables_config=mappings,
    target_catalog=widget("target_catalog", "main"), target_schema=widget("target_schema", "alteryx_migrated"))
converter = SelfCorrectingConverter(engine=engine, max_iterations=int(widget("max_iterations", "3")))
validate = widget("run_sample_validation", "false").lower() == "true"
summaries = batch_export(input_dir, output_dir, formats=formats, converter=converter,
                         spark=globals().get("spark") if validate else None)
assert summaries, "No .yxmd workflows found; set input_dir."
assert all(s["status"] in ("PASS", "WARNING") for s in summaries), summaries
print(json.dumps(summaries, indent=2))
```
